```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/home/abhishek/projects/python_prep/data/sales_data_sample.csv", encoding='windows-1251', usecols=['ORDERNUMBER','QUANTITYORDERED','PRICEEACH','ORDERLINENUMBER','SALES','ORDERDATE','YEAR_ID'], parse_dates=['ORDERDATE'])
filt_df = df[df['YEAR_ID']==2005]
filt_df.head()
```

```python
#result = df.groupby("ORDERLINENUMBER")["SALES"].sum().plot(kind="bar")
result = filt_df.groupby("ORDERLINENUMBER", dropna=False, as_index=False).agg(total_sales=("SALES","sum"),avg_sales=("SALES","mean"))
#high_sales = result[result['total_sales'] > 109359.14]
#print(high_sales.to_string())
sort_result = result.sort_values(by="total_sales", ascending=True).plot(kind="bar")
#sort_result
plt.title("Revenue by store")
plt.ylabel("Total Sales")
plt.xlabel("Store Type")
plt.yticks(rotation=45)
plt.xticks(rotation=45)# Rotates labels so they do not overlap
plt.tight_layout()
plt.show()
#result
#high_sales
sort_result = result.sort_values(by="total_sales", ascending=False)
#sort_result
```
